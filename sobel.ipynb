**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
def sobel_h(img):
    ergebnis = np.zeros((512, 512))

    for z in range(1, 511):
        for s in range(1, 511):
            oben  = img[z-1, s-1] + 2*img[z-1, s] + img[z-1, s+1]
            unten = img[z+1, s-1] + 2*img[z+1, s] + img[z+1, s+1]
            ergebnis[z, s] = (oben - unten) / 8

    return ergebnis
            
def sobel_v(img):
    ergebnis = np.zeros((512, 512))

    for z in range(1, 511):
        for s in range(1, 511):
            links  = img[z-1, s-1] + 2*img[z, s-1] + img[z+1, s-1]
            rechts = img[z-1, s+1] + 2*img[z, s+1] + img[z+1, s+1]
            ergebnis[z, s] = (links - rechts) / 8

    return ergebnis
```

**[Task 2.3.]()** Test your implementations from above:

```python
cells_img_sobelh = sobel_h(cells_img)
cells_img_sobelv = sobel_v(cells_img)

plt.figure()
plt.subplot(1, 2, 1)
plt.imshow(cells_img_sobelh, 'grey')
plt.subplot(1, 2, 2)
plt.imshow(cells_img_sobelv, 'grey')
```

**[Task 2.4.]()** Compute the image gradient magnitude:

```python
staerke = np.sqrt(cells_img_sobelh**2 + cells_img_sobelv**2)

plt.imshow(staerke, cmap="gray")
plt.title("Sobel-Gradientenstärke")
plt.colorbar()
plt.axis("off")
plt.show()
```
